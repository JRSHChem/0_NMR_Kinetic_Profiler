```python

import os
import numpy as np
import pandas as pd
import math
import matplotlib.pyplot as plt
from math import floor
import nmrglue as ng
```

```python
folder = r"xxx"              #input path to nmr folder
file_prefix = "run_"      #input file prefix
file_suffix_num = 10      #input the suffix of the file at first acquisition
N_spectra = 10            #input number of fid files
ref_ppm = 0.08            #input ppm of reference peak
sens = 0.05               #input peak picking sensitivity

ranges = np.array([[8.095, 8.245],[9.945, 10.095]])    #input peakk of interest ranges
nH_ranges = np.array([1,1])                              #input no. of protons in each range
range_name = ["A", "B"]                                 #input range names

cal_range = [-0.075, 0.075]                    #input refrence peak ppm position range
ccal = 0.002                                   #input reference concentration
cal_nH = 12                                    #input reference peak no. of protons

t_start = 300                                      #input time at first acquisotions
t_interval = 600                                   #input intervals between acquisitions

chosen_peak = "A"          #input name (from range_name) of which peak to track

output_folder = r"C:\Users\joesc\Jupyter\test data\nmr_sim\nmr_kinetics_test_data"     #input putput file destination
output_name = "nmr_sim_test.csv" #input output file name

# --- Notes --- #
#file_prefix assumes files are collected in time order, and are labeled as Run_1, Run_1
#
```

```python
#files = sorted([f for f in os.listdir(folder) if f.endswith(".fid")])

all_area_data = []
all_conc_data = []

for n in range(file_suffix_num, file_suffix_num + N_spectra):
    file_name = f"{file_prefix}{n}.fid"
    path = os.path.join(folder, file_name)

    
    dic, data = ng.pipe.read(path)
    udic = ng.pipe.guess_udic(dic, data)
    
    phased_data = ng.process.proc_autophase.autops(data, "acme")
    phased_data_real = phased_data.real
    
    uc = ng.pipe.make_uc(dic, phased_data_real)
    ppm_scale = uc.ppm_scale()
    
    ref_shift_uc = uc('0.00 ppm') - uc(ref_ppm, "ppm")
    data_referenced = np.roll(phased_data_real, ref_shift_uc)
    
    area = ng.analysis.integration.integrate(data_referenced, uc, ranges)
    standard_area = ng.analysis.integration.integrate(data_referenced, uc, cal_range)[0]
    all_area_data.append(area)
    conc = (area/standard_area)*(cal_nH/nH_ranges)*ccal
    all_conc_data.append(conc)
    #print(f"{file_name }area is:{area}")
    
    ratio =area/np.min(area)
    #print(ratio)

    
    peaks = ng.analysis.peakpick.pick(data_referenced, pthres=sens)
    #print(f"peaks: {peaks}")
    
    peak_loc = []
    for i in peaks:
        peak_loc.append(ppm_scale[int(i[0])])
    print(f"peak ppm locations: {peak_loc}")
    
    
    fig = plt.figure(figsize=(10,6))
    ax = fig.add_subplot(1,1,1)
    ax.plot(ppm_scale, data_referenced)
    ax.set_xlabel("Chemical Shift /ppm")
    ax.set_title(file_name)
    plt.gca().invert_xaxis()
    plt.xlim(8, -0.5)
    plt.xticks(np.arange(0, 8, 0.5))
    for p in peak_loc:
        plt.axvline(p, c='C1', ls='--', alpha=1)
    plt.show()

all_time_data = t_start + np.arange(N_spectra) * t_interval

all_area_data = np.array(all_area_data)
all_conc_data = np.array(all_conc_data)
all_time_data = np.array(all_time_data)

col = range_name.index(chosen_peak)
chosen_conc = all_conc_data[:,col]

final_fig = plt.figure(figsize = (10,6))
ax = final_fig.add_subplot(1,1,1)
ax.plot(all_time_data, chosen_conc, "o")
ax.set_xlabel("time /s")
ax.set_ylabel("concentration /M")
plt.show()

kinetics_df = pd.DataFrame(all_conc_data, columns = range_name)
kinetics_df.insert(0,"t_s", all_time_data)
os.makedirs(output_folder, exist_ok = True)
out_file = os.path.join(output_folder, output_name)
kinetics_df.to_csv(out_file, index = False)
print(kinetics_df.head())
```
